### Вопрос 14
Дана колонка со значениями фактора **Продолжительность**.

Проведите нормализацию значений в колонке, при которой каждое значение фактора будет приведено к диапазону $[0, 1]$. В качестве ответа укажите результат нормализации для исходного значения 107. Ответ округлите до 2 знаков после запятой.

```python
duration = np.array([107, 83, 99, 115, 62])

normalized_duration = (duration - duration.min()) / (duration.max() - duration.min())

round(normalized_duration[0], 2)
```

### Вопрос 15
Посчитайте среднее значение фактора **Продолжительность**. В ответе введите среднее, округлённое до двух знаков после запятой.

```python
duration = np.array([33, 98, 72, 117, 75])

duration_mean = duration.mean()

round(duration_mean, 2)
```

### Вопрос 16
Посчитайте разброс значений фактора **Продолжительность**. В ответе введите значение разброса, округлённое до двух знаков после запятой.

```python
duration_std = duration.std()

round(duration_std, 2)
```

### Вопрос 17
Проведите нормализацию значений фактора из предыдущего задания методом стандартизации. В качестве ответа укажите результат нормализации для значения 33. Ответ округлите до 2 знаков после запятой.

```python
normalized_duration = (duration - duration_mean) / duration_std

round(normalized_duration[0], 2)
```

### Вопрос 22
Соотнесите конкретную проблему, которая может возникнуть при создании модели машинного обучения, с этапом обработки факторов, способным помочь её решить.

*Ответ:*

1. разные факторы вносят в итоговое предсказание модели вклад, пропорциональный порядку величин, в которых они измеряются, а не значимости этих факторов. Это приводит к трудностям с интерпретацией модели: ***нормализация факторов;***

2. в имеющихся для решения задачи данных представлено очень много факторов, часть из которых неинформативны или излишни. Из-за большого количества факторов модель обучается и работает хуже, чем могла бы: ***отбор факторов;***

3. в данных присутствуют содержательные факторы, которые хочется использовать при построении модели, однако факторы являются категориальными, в частности принимают строковые значения, то есть напрямую при обучении модели их не задействовать: ***кодирование категориальных факторов;***

4. численный фактор имеет большой диапазон значений, в то время как сами значения сгруппированы в нескольких конкретных интервалах. Необходимо явно указать модели, что содержательных групп значений фактора всего несколько, — это упростит её работу и интерпретацию: ***дискретизация численных факторов.***
