### Вопрос 12
Дан следующий набор данных. С помощью метода МНК постройте по этим данным модель линейной регрессии с двумя параметрами: свободным коэффициентом $w_0$ и весом $w_1$, который соответствует фактору $x$.

В качестве ответа введите через пробел значения параметров $w_0$ и $w_1$, округлённые до двух знаков после запятой. 

```python
x = np.array([1, 2, 4, 5])
y = np.array([3, 2, 4, -1])

X = np.vstack([np.ones(len(x)), x]).T

w = np.linalg.inv(X.T @ X) @ X.T @ y

w0, w1 = w.round(2)

print(f"w0 = {w0}, w1 = {w1}")
```

Также дан набор данных. Ниже вам будет нужно поэтапно вычислить среднеквадратичную ошибку предсказания модели на представленном наборе данных по формуле:

$$
\frac{1}{n} \| y - X w \|^2
$$

В этой формуле:

- $y$ — вектор реальных значений предсказываемой величины;
- $X$ — матрица параметров модели, в которую слева был добавлен столбец из единиц, соответствующий искусственному фактору $x_0$;
- $w$ — вектор параметров модели;
- $n$ — число объектов в данных.

```python
X = np.array([[2, 3], [4, -1], [-3, 1]])
y_true = np.array([1, -2, 3])
w = np.array([-1, 2, -3])

X = np.hstack([np.ones((len(X), 1)), X])
```

1. Вычислите результат умножения матрицы $X$ на вектор $w$.

```python
y_pred = X @ w

y_pred
```

2. Вычислите результат вычитания из вектора $y$ вектора $Xw$.

```python
diff = y_true - y_pred

diff
```

3. Вычислите длину вектора $y−Xw$ по формуле вычисления длины вектора в многомерном пространстве. В ответе укажите $\frac{1}{n} \| y - X w \|^2$, то есть искомую среднеквадратичную ошибку. Ответ округлите до двух знаков после запятой.

```python
mse = np.mean(diff ** 2)

round(mse, 2)
```

```python
X = np.array([[1, -1], [-1, 1], [1, 1]])
y_true = np.array([1, -2, 2])

X = np.hstack([np.ones((len(X), 1)), X])
```

$$

### Вопрос 30
Была проведена дискретизация фактора $x$. Множество его значений разбили на два интервала: $[0, 2)$ и $[2, 4)$.

Получившийся категориальный фактор был переведён в численный за счёт добавления в данные двух новых колонок: $[0, 2)$ и $[2, 4)$. Значение 1 в какой-то из них соответствует тому, что значение фактора $x$ соответствующего объекта находилось в этом интервале.

Постройте модель линейной регрессии по набору данных, в которых значение фактора $x$ было дискретизировано.

Лучшее, что может предсказать модель на каждом интервале, — это среднее значение предсказываемой характеристики для объектов, попавших в интервал.

В качестве ответа укажите среднеквадратичную ошибку предсказания модели. Значение ошибки округлите до 2 знаков после запятой.

1. Подготовка данных

```python
import pandas as pd
from pandas.api.types import CategoricalDtype

data = pd.DataFrame({"x": [1.5, 1.8, 2.5, 3.0], "y": [2, 8, 3, -1]})
```

2. Дискретизация численного фактора на равные интервалы.

```python
bins = [0, 2, 4]
labels = ["[0, 2)", "[2, 4)"]

interval_type = CategoricalDtype(categories=labels, ordered=True)

data["x_interval"] = pd.cut(data["x"], bins=bins, labels=labels, right=False).astype(
    interval_type
)
```

3. Вычисление предсказанного среднего значения по интервалам.

```python
interval_means = data.groupby("x_interval", observed=True)["y"].mean()

data["y_pred"] = data["x_interval"].map(interval_means).astype(float)
```

4. Вычисление метрики оценки MSE.

```python
mse = np.mean((data["y"] - data["y_pred"]) ** 2)

round(mse, 2)
```
