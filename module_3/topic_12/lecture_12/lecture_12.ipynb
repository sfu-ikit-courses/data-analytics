### Вопрос 9
В предыдущем задании мы решали более простую задачу, чем та, которую решают люди, занимающиеся разметкой изображений. Обычно по умолчанию неизвестно, какие объекты есть на изображении, и задача специалиста по разметке данных заключается в том, чтобы правильно выделить и дать именования необходимым объектам. Будем считать, что в среднем разметка одного изображения в специальной программе, которая делает обработку большого числа картинок более удобной, занимает порядка 2 минут. В день специалист по разметке данных работает 9 часов, один из которых тратит на обед. Для простоты вычислений будем считать, что в месяце 22 рабочих дня. Сколько лет потребуется 10 специалистам по разметке данных, чтобы разметить все изображения в базе размером 14 000 000 картинок?

В качестве ответа введите целое число лет. Если ответ получился дробным, то округлите его вниз.

```python
number_of_employees = 10
working_hours_per_day = 8
time_per_image_minutes = 2
working_days_per_month = 22
total_images_to_process = 14_000_000

images_per_employee_per_day = (working_hours_per_day * 60) / time_per_image_minutes

total_images_per_month = (
    images_per_employee_per_day * number_of_employees * working_days_per_month
)

total_years_needed = int(total_images_to_process / total_images_per_month / 12)

print(
    f"Общее количество лет, необходимое для обработки всех изображений: {total_years_needed}"
)
```

### Вопрос 17
С помощью модели линейной регрессии для 5 учеников были предсказаны итоговые оценки. Оцените качество модели как среднее отклонение предсказанных оценок от реальных.

Ответ округлите до двух знаков после запятой.

```python
import numpy as np

real_grades = np.array([12, 9, 16, 20, 18])
predicted_grades = np.array([11, 6, 17, 18, 15])

absolute_errors = np.abs(real_grades - predicted_grades)

mae = np.mean(absolute_errors)

print(f"Среднее отклонение (MAE): {mae:.2f}")
```
