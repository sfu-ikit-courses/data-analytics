### Вопрос 1
Дана проходимость магазина бытовой техники за неделю:

- понедельник: 50 клиентов;
- вторник: 60 клиентов;
- среда: 40 клиентов;
- четверг: 47 клиентов;
- пятница: 53 клиента.

Проверим гипотезу на уровне значимости $\alpha=0.05$ о том, что каждый будний день в магазин приходит одинаковое количество покупателей.

```python
observed = np.array([50, 60, 40, 47, 53])
```

1. Нулевая гипотеза: каждый день в магазин заходит ... количество покупателей.

*Ответ:* разное

2. Рассчитайте теоретические частоты для посещаемости магазина. Укажите получившееся значение для каждого дня.

```python
total = sum(observed)
n = len(observed)

total // n
```

3. Значение критерия согласия хи-квадрат округленное до 2 знаков после запятой:

```python
expected = np.array([50] * 5)

chi = ((observed - expected) ** 2 / expected).sum()

round(chi, 2)
```

4. Число степеней свободы:

```python
dof = n - 1

dof
```

5. Критическое значение критерия для $\alpha=0.05$ и найденного числа степеней свободы, округленное до 2 знаков после запятой:

```python
alpha = 0.05

chi2_critical = chi2.ppf(1 - alpha, dof)

round(chi2_critical, 2)
```

6. С использованием библиотеки `scipy.stats` рассчитайте значение *pvalue* округленное до 2 знаков после запятой.

```python
chi2_stat, p_value = chisquare(observed, expected)

round(p_value, 2)
```

*Вывод:* Нулевая гипотеза принимается.

## Часть 2

Для успешного прохождения теста вам потребуется воспользоваться [данными](./StarWars.csv) по серии фильмов «Звездные войны», самостоятельно загрузив их в *Python*.

Описание данных
- **respondentID** — идентификационный номер респондента
- **seen** — ответ на вопрос, видел ли респондент хотя бы один фильм из серии «Звездных войн» (да/нет)
- **fan** — ответ на вопрос, считает ли респондент себя фанатом серии фильмов (да/нет)
- **gender** — пол респондента (М/Ж)
- **age_group** — возрастная группа, к которой принадлежит респондент (18-29, 30-44, 45-60, > 60)
- **income** — доход респондента в долларах США (0 - 24.999, 25.000 - 49.999, 50.000 - 99.999, 100.000 - 149.999, 150.000+)
- **loc** — регион проживания респондента (в США)
- **yoda** — отношение респондента к персонажу серии (Йоде)
- **princess_leia_organa** — отношение респондента к персонажу серии (принцессе Лее Органе)

```python
survey = pd.read_csv("StarWars.csv", sep=";")
survey.head(1)
```

*Обработка пропущенных значений полей **gender** и **fan**:*

```python
survey = survey.dropna(subset=["gender", "fan"])
```

### Вопрос 1
Обратим внимание на переменные **gender** и **fan**. Какую гипотезу мы можем протестировать, используя хи-квадрат?


*Ответ:* Распределение наблюдений по категориям выбранных переменных независимо.

### Вопрос 2
Постройте таблицу сопряженности для переменных **gender** и **fan**. Укажите кол-во женщин, являющихся фанатами серии фильмов «Звездные войны».

```python
contingency_table = pd.crosstab(survey["gender"], survey["fan"])

contingency_table.loc["Ж", "да"]
```

### Вопрос 3
Предположим, что мы тестируем гипотезу о том, что распределение наблюдений по категориям выбранных нами переменных независимо. Чему будет равно количество степеней свободы, рассчитанное по построенной в предыдущем задании таблице сопряженности?

```python
r, c = contingency_table.shape

dof = (r - 1) * (c - 1)

dof
```

### Вопрос 4
Что означает равенство уровня значимости одному проценту?

*Ответ:* Вероятность ошибки первого рода равна 1%.

### Вопрос 5
Чему будет равно критическое значение критерия хи-квадрат для указанных выше параметров (уровня значимости 5% и верного числа степеней свободы), при котором мы сможем отвергнуть нулевую гипотезу?

```python
alpha = 0.05

chi2_critical = chi2.ppf(1 - alpha, dof)

round(chi2_critical, 2)
```

### Вопрос 6
Чему равна оценка вероятности того, что человек НЕ является фанатом серии фильмов? Ответ округлите до двух знаков после запятой.

```python
not_fan_count = len(survey[survey["fan"] == "нет"])
total_count = len(survey)

probability_not_fan = not_fan_count / total_count

round(probability_not_fan, 2)
```

### Вопрос 7
Чему равна теоретическая частота для женщин, которые НЕ являются фанатами серии фильмов? Ответ округлите до целого.

```python
sum_women = len(survey[survey["gender"] == "Ж"])
sum_no_fans = len(survey[survey["fan"] == "нет"])

E_women_fans = (sum_women * sum_no_fans) / total_count

round(E_women_fans)
```

### Вопрос 7
Чему равен критерий хи-квадрат? Ответ округлите до двух знаков после запятой.

```python
observed = contingency_table.values

row_sums = observed.sum(axis=1).reshape(-1, 1)
col_sums = observed.sum(axis=0).reshape(1, -1)
total = observed.sum()
expected = row_sums @ col_sums / total

chi = ((observed - expected) ** 2 / expected).sum()

round(chi, 2)
```

### Вопрос 8
С помощью библиотеки `scipy.stats`  рассчитайте значение *pvalue*. Ответ округлите до 4 знаков после запятой.

```python
p_value = chi2.sf(chi, dof)

round(p_value, 4)
```
