### 1. Назначение мер центральной тенденции. Примеры использования в повседневной жизни.

В попытке найти единственное значение, которое бы лучше всего репрезентовало данные, было достаточно типичным, мы вводим понятие ***центральной тенденции***. Поиск значения, выражающего центральную тенденцию, — это поиск такого значения, которое находится в центре распределения.

Примеры использования:

- средний чек покупательской корзины (для определения минимальной зарплаты, например);
- среднее телосложение людей для пошива одежды;
- среднее количество минут на то, чтобы дойти от дома до университета.

### 2. Cреднее арифметическое (определение). Как определить  среднее арифметическое генеральной совокупности? Как определить  среднее арифметическое выборки?

***Среднее арифметическое*** — число, равное сумме всех чисел выборки или генеральной совокупности, деленной на их количество.


Пусть $X_i$ — $i$-ое значение генеральной совокупности, тогда *формула среднего арифметического генеральной совокупности:*

$$
\mu = \frac{\sum X_i}{N}
$$

Пусть $x_i$ — $i$-ое значение выборки, тогда *формула среднего арифметического выборки:*

$$
M = \frac{\sum x_i}{n}
$$

### 3. Свойства среднего арифметического и применение.

**Свойства:**

- *балансировочная точка*. Это означает, что сумма отклонений от среднего равна нулю;
- *чувствительность к аномалиям*. Среднее арифметическое может значительно измениться при добавлении очень больших или очень малых значений.

**Применение:**

Использовать среднее арифметическое имеет смысл когда переменные - непрерывные количественные и когда распределение симметричное (нет аномальных значений). Примерами таких данных могут быть: ошибка измерения, оценки за экзамен школьников из одного класса, температура.

### 4. Измерение центральной тенденции (среднего, медианы и моды)

***Медиана*** — это точка ниже и выше которой находится ровно по 50% значений в распределении. Чтобы найти медиану, нам нужно упорядочить значения переменной от меньшего к большему и найти то значение, которое делит распределение пополам.

***Мода*** — это значение переменной с самой большой частотой, т.е. самое популярное значение переменной.

В Pandas для вычисления измерений центральной тенденции существует соответствующие функции.

```python
np.random.seed(42)

ages = pd.Series(np.random.randint(20, 30, 20))

print("Среднее арифметическое:", round(ages.mean(), 2))
print("Медиана:", round(ages.median()))
print("Мода:", ", ".join(map(str, ages.mode())))
```

### 5. Меры вариативности и их назначение. Вычисление размаха, интерквартильного размаха, дисперсии, стандартного отклонения.

***Меры вариативности (разброса)*** представляют собой количественные меры, выражающие различия между значениями в распределении. Более формально — это степень индивидуальных отклонений значений от центральной тенденции.

Они помогают нам понять, насколько наши данные разнообразны и насколько типичной для распределения является случайно взятая оценка.

***Размах (R)*** — это разница между самым большим и самым маленьким значением переменной. Интерпретация размаха: все значения переменной находятся в диапазоне размаха.

*Формула размаха генеральной совокупности и выборки:*

$$
R = x_{max} - x_{min}
$$

***Квартили (Q)*** — это значения, которые делят распределение на четверти

***Интерквартильный размах (IQR)*** — интервал значений признака, содержащий центральные 50% наблюдений распределения, то есть интервал между первым и третьим квартилем. Используется в паре с медианой. Устойчива к аномалиям.

*Формула интерквартильного размаха генеральной совокупности и выборки:*

$$
IQR = Q_3 - Q_1
$$

***Отклонение от среднего*** — это разница между значением переменной и ее средним арифметическим.

Найдем среднее арифметическое всех отклонений от среднего. Мы знаем, что оно равно 0, поэтому найдем квадрат всех отклонений от среднего. Это называется дисперсией.

***Дисперсия*** — среднее квадратов отклонений от среднего арифметического распределения.

*Формула дисперсии генеральной совокупности:*

$$
\sigma^2 = \frac{\sum (X_i - \mu)^2}{N}
$$

*Формула дисперсии выборки:*

$$
s^2 = \frac{\sum (x_i - M)^2}{n - 1}
$$

Выражать среднее отклонение в квадрате неудобно. Возьмем корень от дисперсии и получим среднеквадратичное отклонение.

***Среднеквадратичное отклонение (стандартное отклонение)*** — квадратный корень дисперсии. Мера, которая определяет среднее отклонение от среднего арифметического распределения.

*Формула cреднеквадратичного отклонения генеральной совокупности:*

$$
\sigma = \sqrt{\frac{\sum (X_i - \mu)^2}{N}}
$$

*Формула cреднеквадратичного отклонения выборки:*

$$
s = \sqrt{\frac{\sum (x_i - M)^2}{n - 1}}
$$

В Pandas для вычисления измерений мер вариативности существует соответствующие функции.

```python
np.random.seed(42)

values = pd.Series(np.random.randint(20, 30, 20))

print("Размах:", values.max() - values.min())
print("IQR:", values.quantile(0.75) - values.quantile(0.25))
print("Дисперсия (генеральная совокупность):", round(values.var(ddof=0), 2))
print("Дисперсия (выборка):", round(values.var(), 2))
print(
    "Стандартное отклонение (генеральная совокупность):", round(values.std(ddof=0), 2)
)
print("Стандартное отклонение (выборка):", round(values.std(), 2))
```

### 6. z-оценка (определение и  назначение). Измерение z-оценки для генеральной совокупности и для выборки. Выражение среднего и среднеквадратичного отклонения через z-оценку.

***Стандартизированная оценка (z-оценка)*** — это относительная мера, которая показывает, на сколько среднеквадратичных отклонений наблюдаемое значение отличается от среднего значения распределения. Знак z-оценки показывает, находится ли значение левее среднего (–) или правее среднего (+).

С помощью z-оценок мы можем стандартизировать распределение — выразить все значения в терминах дистанции от среднего.

*Формула z-оценки генеральной совокупности:*

$$
z = \frac{X - \mu}{\sigma}
$$

*Формула z-оценки выборки:*

$$
z = \frac{x - M}{s}
$$

Зная z-оценку, среднее и среднеквадратичное отклонение, можно легко восстановить и оригинальное значение.

*Для генеральной совокупности:*

$$
X = \mu + z \cdot \sigma
$$

*Для выборки:*

$$
X = M + z \cdot s
$$

### 7. Стандартизированное распределение и его свойства.

Если мы превратим все значения переменной распределения в z-оценки, то мы получим ***z-распределение или стандартизированное распределение***. Такое распределение сохранит форму оригинального распределения, но среднее в нем будет равно нулю, а среднеквадратичное отклонение — единице.

**Свойства:**

Через z-оценки можно описать пропорции нормального распределения: в интервале от минус одного до плюс одного среднеквадратичного отклонения располагается примерно 68% выборки — то есть самые типичные для нее значения.

### 8. Выбросы (определение и  назначение).

***Выброс*** — это аномальное значение в данных, которое значительно отличается от значения, выраженного мерой центральной тенденции.

Понятие выброса может быть субъективным, но есть некоторые принятые конвенции: значения отличающиеся от среднего на 3 и больше среднеквадратичных отклонений называют *сильными* выбросами, а от 2 до 3 отклонений – *умеренными*.

**Назначение:**

- могут нести какую-то интересную информацию, которая поможет исследованию (если зарплата одного из выпускников сильно выше большинства, можно узнать, чем отличается этот выпускник от других и сделать определенные выводы);
- выбросы могут повлиять на расчеты метрик, поэтому своевременное нахождение таких аномалий поможет в дальнейшем.

### 9. Исключение выбросов с помощью Python.

**Способ 1:** Будем считать выбросами все, что находится ниже 1-й квартили на 1,5 интерквартильных размаха и выше 3-й квартили на 1,5 интерквартильных размаха.

```python
np.random.seed(42)

views = np.random.normal(loc=50, scale=10, size=100)
views = pd.Series(np.append(views, [150, 300, 450]))

q1 = views.quantile(0.25)
q3 = views.quantile(0.75)
iqr = q3 - q1

lower_iqr = q1 - 1.5 * iqr
upper_iqr = q3 + 1.5 * iqr

filtered_iqr = views[views.between(lower_iqr, upper_iqr)]

print(
    f"Диапазон без выбросов (по IQR): от {round(filtered_iqr.min(), 1)} до {round(filtered_iqr.max(), 1)}"
)
```

**Способ 2 (Использование среднеквадратического отклонения):** чаще всего выбросами считаются наблюдения, находящиеся на расстоянии 3 и более среднеквадратических отклонения от среднего (правило 3-х сигм).

```python
mean = views.mean()
std = views.std()

lower_std = mean - 3 * std
upper_std = mean + 3 * std

filtered_std = views[views.between(lower_std, upper_std)]

print(
    f"Диапазон без выбросов (по 3 std): от {round(filtered_std.min(), 1)} до {round(filtered_std.max(), 1)}"
)
```
