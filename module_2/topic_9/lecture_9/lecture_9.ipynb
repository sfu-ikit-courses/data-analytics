### Вопрос 2
Дана [следующая таблица](./uniform_and_not_data.csv) в формате *csv*. В таблице для каждого наблюдения представлены значения пяти факторов: *A*, *B*, *C*, *D* и *E*.

Определите, какие факторы в данных имеют распределение, которое отличается от равномерного.

```python
data = pd.read_csv("uniform_and_not_data.csv")


def is_uniform(data: pd.Series, skew_tol=0.1, kurt_tol=0.2) -> bool:
    skewness = data.skew()
    kurtosis = data.kurt()

    is_skew_ok = abs(skewness) <= skew_tol
    is_kurt_ok = abs(kurtosis + 1.2) <= kurt_tol

    return is_skew_ok and is_kurt_ok


for column in data.columns:
    if not is_uniform(data[column]):
        print(column)
```

### Вопрос 22
Дан массив значений зарплат работников: $[18000, 32000, 102000, 54000, 88000]$

После проведения нормализации массив принял вид (значения округлены до 2 знаков после запятой): $[0. , 0.17, 1. , 0.43, ...]$

В ответе укажите пропущенное значение.

```python
from sklearn.preprocessing import MinMaxScaler

salaries = pd.DataFrame([[18000], [32000], [102000], [54000], [88000]])

normal_scaler = MinMaxScaler()
normalized_salaries = normal_scaler.fit_transform(salaries).round(2)

normalized_salaries[-1]
```
