### 1. *t*-статистика и *t*-распределение: понятие и назначение.

***t — статистика*** — критерий, который используется вместо *z*-оценки для проверки гипотез о среднем для генеральной совокупности, когда неизвестно стандартное отклонение для генеральной совокупности.

Для вычисления *t*-статистики используют формулу:

$$
t = \frac{M - \mu}{s_m}
$$

Где:
- $M$ — выборочное среднее,
- $\mu$ — гипотетическое среднее,
- $s_m$ — оценка стандартной ошибки.


Для вычисления оценки стандартной ошибки используется следующая формула:

$$
s_m = \frac{s}{\sqrt{n}}
$$

Где:
- $s$ — стандартное отклонение выборки,
- $n$ — размер выборки.

***t-распределение*** — это набор всех значений *t*-статистик, вычисленных для всех возможных случайных выборок определенного размера. *t*-распределение стремится к нормальному распределению с увеличением выборки.


### 2. Одновыборочный *t*-тест: применение и алгоритм проверки гипотез.

***Одновыборочный t-тест*** используется для проверки, отличается ли среднее значение выборки от заданного значения (например, является ли рост 170 см средним для женщин).

*Алгоритм проверки гипотез:*

1. находим наблюдаемое значение *t*-статистики по формуле:

$$
t_{наблюдаемое} = \frac{M - \mu}{s_m}
$$


2. по уровню значимости и количеству степеней свободы определяем критические значения *t*-статистики по специальной таблице;

3. если наблюдаемое значение не попадает в критическую область — не отвергаем нулевую гипотезу. Если наблюдаемое значение попадает в критическую область — отвергаем нулевую гипотезу в пользу альтернативной.

### 3. Реализация одновыборочного *t*-теста в *Python*.

Исследуем средний балл студентов на экзамене по математике.

**Нулевая гипотеза $H_0$:** средний балл студентов на экзамене = 70

**Альтернативная гипотеза $H_1$:** средний балл студентов на экзамене $\neq$ 70

**Данные (баллы):** 72, 65, 80, 85, 70, 68, 74, 79.

```python
data = np.array([72, 65, 80, 85, 70, 68, 74, 79])
mu_0 = 70

t_stat, p_value = stats.ttest_1samp(data, mu_0)

print(f"t-статистика: {t_stat:.3f}")
print(f"p-значение: {p_value:.3f}")

alpha = 0.05

if p_value < alpha:
    print("Отклоняем нулевую гипотезу: средний балл отличается от 70.")
else:
    print("Нет оснований отклонять нулевую гипотезу: средний балл ≈ 70.")
```

### 4. Задание

Решите 2 способами (проведя свои расчёты в *Python* и с помощью библиотеки `scipy.stats`) задачу: даны баллы за экзамен по английскому языку, проверьте гипотезу о равенстве среднего балла всех учеников значению 6,5. Сравните результаты решений.

Баллы: 8, 6, 4, 6, 8, 9, 7, 8.

```python
scores = np.array([8, 6, 4, 6, 8, 9, 7, 8])
mu_0 = 6.5
```

Рассчитаем вручную:

```python
n = len(scores)
mean_x = np.mean(scores)
std_x = scores.std(ddof=1)

t_stat_manual = (mean_x - mu_0) / (std_x / np.sqrt(n))
df = n - 1
p_value_manual = 2 * (1 - stats.t.cdf(abs(t_stat_manual), df))

round(t_stat_manual, 3), round(p_value_manual, 3)
```

Теперь с помощью библиотеки `scipy.stats`:

```python
t_stat_lib, p_value_lib = stats.ttest_1samp(scores, mu_0)

round(t_stat_lib, 3), round(p_value_lib, 3)
```

**Вывод:** Так как *p_value* $>$ 0.05, в обоих случаях нет оснований отвергать нулевую гипотезу. Средний бал действительно приближенно равен 6.5.

Таким образом, результаты практически не отличаются (расчет *p_value* очень приближенный). Вероятно в библиотеке используется тот же алгоритм, который мы описали вручную.

### 5. Парный *t*-тест: применение и алгоритм проверки гипотез.

***Парный t-тест*** используется для проверки, изменились ли значения одной и той же группы после какого-то воздействия. Например, проверка значений до и после эксперимента для одной и той же группы участников.

*Алгоритм проверки гипотез:*

1. находим разницы *D* между вторым значением и первым для каждого объекта;

2. находим наблюдаемое значение *t*-статистики для полученных разниц по формуле:

$$
t_{наблюдаемое} = \frac{M_D - \mu_D}{s_{m_D}}
$$

3. по уровню значимости и количеству степеней свободы определяем критические значения *t*-статистики по таблице;

4. если наблюдаемое значение не попадает в критическую область — не отвергаем нулевую гипотезу. Если наблюдаемое значение попадает в критическую область — отвергаем нулевую гипотезу в пользу альтернативной.

### 6. Реализация парного *t*-теста в *Python*.

Исследуем, улучшились ли баллы студентов после прохождения дополнительного курса по математике.

**Нулевая гипотеза $H_0$:** разница в баллах до и после курса = 0 (то есть курс не повлиял на результаты).

**Альтернативная гипотеза $H_1$:** разница в баллах до и после курса $\neq$ 0 (то есть курс повлиял на результаты).

**Данные (Баллы):**
- до курса: 60, 62, 58, 65, 55, 64, 60, 70;
- после курса: 61, 61, 63, 67, 60, 68, 65, 70.

```python
before = np.array([60, 62, 58, 65, 55, 64, 60, 70])
after = np.array([61, 61, 63, 67, 60, 68, 65, 70])


t_stat, p_value = stats.ttest_rel(before, after)

print(f"t-статистика: {t_stat:.3f}")
print(f"p-значение: {p_value:.3f}")

alpha = 0.05
if p_value < alpha:
    print("Отклоняем нулевую гипотезу: курс улучшил результаты.")
else:
    print("Нет оснований отклонять нулевую гипотезу: курс не повлиял на результаты.")
```

### 8. Двухвыборочный *t*-тест: применение и алгоритм проверки гипотез.

***Двухвыборочный t-тест*** используется для проверки, отличаются ли средние значения в двух независимых группах.

***Алгоритм проверки гипотез:***

1. находим средние для обеих выборок;

2. находим объединенное стандартное отклонение:

$$
s_{M_1 - M_2} = \sqrt{\frac{s^2_1}{n_1} + \frac{s^2_2}{n_2}}
$$

3. вычисляем значение наблюдаемой *t*-статистики:

$$
t_{наблюдаемое} = \frac{(M_1 - M_2) - (\mu_1 - \mu_2)}{s_{M_1 - M_2}}
$$

4. по уровню значимости и количеству степеней свободы определяем критические значения *t*-статистики по таблице;

5. если наблюдаемое значение не попадает в критическую область — не отвергаем нулевую гипотезу. Если наблюдаемое значение попадает в критическую область — отвергаем нулевую гипотезу в пользу альтернативной.

### 9. Реализация двухвыборочного *t*-теста в *Python*.

Исследуем, отличаются ли баллы студентов в двух разных группах на экзамене по математике.

**Нулевая гипотеза $H_0$:** средний балл студентов в двух группах одинаков.

**Альтернативная гипотеза $H_1$:** средний балл студентов в двух группах различен.

**Данные (Баллы):**
- Группа 1: 75, 80, 85, 90, 70, 95, 60, 85;
- Группа 2: 70, 65, 80, 85, 75, 90, 80, 70.

*Входные данные:*

```python
group_1 = np.array([75, 80, 85, 90, 70, 95, 60, 85])
group_2 = np.array([70, 65, 80, 85, 75, 90, 80, 70])
```

*Проверка на соотношение дисперсий:*

```python
var_group_1 = group_1.var(ddof=1)
var_group_2 = group_2.var(ddof=1)

max_var = max(var_group_1, var_group_2)
min_var = min(var_group_1, var_group_2)

var_ratio = max_var / min_var

equal_var = var_ratio < 4
```

*Основной блок решения:*

```python
t_stat, p_value = stats.ttest_ind(group_1, group_2, equal_var=equal_var)

print(f"t-статистика (двухвыборочный t-тест): {t_stat:.3f}")
print(f"p-значение (двухвыборочный t-тест): {p_value:.3f}")

alpha = 0.05
if p_value < alpha:
    print("Отклоняем нулевую гипотезу: баллы в группах различаются.")
else:
    print("Нет оснований отклонять нулевую гипотезу: баллы в группах одинаковы.")
```

### 10. Задание

Решите 2 способами (проведя свои расчёты в *Python* и с помощью библиотеки `scipy.stats`) задачу: даны данные для 2 школ английского языка. Проверьте гипотезу: студенты в среднем сдают экзамен одинаково. Сравните результаты решений.

Баллы для 1 школы: 5, 2, 7, 7, 6, 5, 4, 4.

Баллы для 2 школы: 7, 9, 6, 9, 6, 8, 4, 7.

```python
school_1 = np.array([5, 2, 7, 7, 6, 5, 4, 4])
school_2 = np.array([7, 9, 6, 9, 6, 8, 4, 7])
```

Рассчитаем вручную:

```python
mean1 = school_1.mean()
mean2 = school_2.mean()

var1 = school_1.var(ddof=1)
var2 = school_2.var(ddof=1)

n1, n2 = len(school_1), len(school_2)

se = np.sqrt(var1 / n1 + var2 / n2)

t_stat_manual = (mean1 - mean2) / se

df = n1 + n2 - 2

p_value_manual = 2 * (1 - stats.t.cdf(abs(t_stat_manual), df))

print(f"t-статистика: {t_stat_manual:.3f}")
print(f"p-value: {p_value_manual:.3f}")
```

Теперь с помощью библиотеки `scipy.stats`:

```python
t_stat_lib, p_val_lib = stats.ttest_ind(school_1, school_2, equal_var=True)

print(f"t-статистика: {t_stat_lib:.3f}")
print(f"p-value: {p_val_lib:.3f}")
```

**Вывод:** Так как *p_value* $<$ 0.05, в обоих случаях отвергаем нулевую гипотезу. Студенты из разных школ в среднем сдают экзамены по-разному.

Таким образом, результаты практически не отличаются (расчет *p_value* очень приближенный). Вероятно в библиотеке используется тот же алгоритм, который мы описали вручную.
